# 比較演算子

- 等しい（`==`）、〇〇以上（`>=`）、〇〇未満（`<`）など、左辺と右辺の値を比較するための演算子を**比較演算子**という。  
- 比較の結果は、`True`（真）または`False`（偽）で返される。  
- 変数に値を代入する場合は`=`を使うが、値の比較には`==`を使うことに注意する。  

| 比較演算子 | 意味       |
|:----------:|:----------|
| `<`        | 小さい     |
| `>`        | 大きい     |
| `<=`       | 以下       |
| `>=`       | 以上       |
| `==`       | 等しい     |
| `!=`       | 等しくない |

## 数値の比較

````python
num1 = 10

# 等しい
print("判定1: ", num1 == 10)

# 等しくない
print("判定2: ", num1 != 10)

# 大きい
print("判定3: ", num1 > 10)

# 小さい
print("判定4: ", num1 < 10)

# 以上
print("判定5: ", num1 >= 10)

# 以下
print("判定6: ", num1 <= 10)
````

# 論理演算子
- 2つ以上の比較条件を組み合わせて判定するときに使う。  
- 結果は `True`（真）または `False`（偽）で返される。  

| 論理演算子 | 意味           |
|:----------:|:--------------|
| `and`      | 両方の条件を満たす場合に真となる |
| `or`       | どちらか一方または両方の条件を満たす場合に真となる |
| `not`(参考)| 条件を満たさない場合に真となる（論理否定） |

## ２つの比較条件の組み合わせ

````python
num1 = 10
num2 = 20

# 条件1と条件2の両方を満たすか
print("判定1は", num1 >= 10 and num2 <= 10)

# 条件1と条件2のどちらかを満たすか
print("判定2は", num1 >= 10 or num2 <= 10)
````

## ３つの比較条件の組み合わせ

````python
num1 = 10
num2 = 20
num3 = 30

# 条件1と条件2と条件3を同時に満たすか
print("判定1は", num1 >= 10 and num2 <= 10 and num3 == 10)

# 条件1と条件2と条件3のいずれかを満たすか
print("判定2は", num1 >= 10 or num2 <= 10 or num3 == 10)
````

# 制御構文
- プログラムに電卓以上の高度な機能を持たせるためには、**制御構文**が必要である。  
- **制御構文**とは、プログラムの実行順序を制御するための仕組みである。  
- プログラムの実行は、**順次処理**、**条件分岐（選択）**、**繰り返し（反復・ループ）**の3つの制御構文の組み合わせで表現される。  
- これら3つを組み合わせることで、あらゆる実行順序を記述できる。

# 条件分岐（選択）

## 条件文（if文）
- 条件分岐（選択）には`条件文（if文）`を用いる
- if文を用いることで、**「もし〇〇だったら、処理A。そうでなく△△だったら、処理B。」**のように、条件によって処理を変更できる
----
- **条件１つの処理（if文）**
  ```python
  if 条件式1:
      条件式1がTrue（真）ならば、ここに記述された処理を行う
  ```
  - ``条件式の後に`:`(コロン)の付け忘れに注意``
----

- **条件1つ＋それ以外の場合の処理（if else文）**
  ```python
  if 条件式1:
      条件式1がTrue（真）ならば、ここに記述された処理を行う
  else:
      条件式1がFalse（偽）の場合に実行する処理
  ```
  - ``elseの後に`:`(コロン)の付け忘れに注意``  
----


- **複数条件＋それ以外の場合の処理（if elif else文）**
  ```Python
  if 条件式1:
    条件式1がTrue(真)ならば、ここに記述された処理を行う
  elif 条件式2:
    条件式1がFalse(偽)で、条件式2がTrue(真)ならば、ここに記述された処理を行う
  else:
    全ての条件式がFalse(偽)ならば、ここに記述された処理を行う
  ```
----  
- **注意事項**<br>
  - if文の処理を記述する際は、**空白（半角スペース）を使って字下げ（インデント）を揃える**必要がある。通常は半角スペース4つ分。  
  - 毎回スペースキーを4回押すのは面倒なので、**Tabキー**を使うと自動的にインデントが調整される。  

## 条件１つの処理（if文）
変数numが100未満なら「100未満です」と出力

````python
num = 50

if num < 100:
    print("100未満です")
````

## 条件1つ＋それ以外の場合の処理（if else文）
変数numが100未満なら「100未満です」、それ以外は「100以上です」と出力

````python
num = 101

if num < 100:
    print("100未満です")
else:
    print("100以上です")
````

## 条件2つ＋それ以外の場合の処理（if elif else文）
変数numが100未満なら「100未満です」、<br>100以上200未満なら「100以上200未満です」、<br>それ以外は「200以上です」と出力

````python
num = 150

if num < 100:
    print("100未満です")
elif num < 200:
    print("100以上200未満です")
else:
    print("200以上です")
````

## 条件3つ＋それ以外の場合の処理（if elif else文）
変数numが100未満なら「100未満です」、<br>100以上200未満なら「100以上200未満です」、<br>200以上300未満なら「100以上300未満です」、<br>それ以外は「300以上です」と出力

````python
num = 150

if num < 100:
    print("100未満です")
elif num < 200:
    print("100以上200未満です")
elif num < 300:
    print("100以上300未満です")
else:
    print("300以上です")
````

## 論理演算子（and）を用いた条件文

トランプのカードが「ダイア（♦）」、かつ数字が10以上の場合は10点を加算し、それ以外の場合は2点を加算するプログラムを作成する。最初にポイントは100点を持っているものとする。

````python
points = 100 # ポイント、初期値

number = 10          # トランプの番号
mark   = "diamond"  # トランプのマーク

if number >=10 and mark == "diamond":
    points = points + 10
else:
    points = points + 2

print("ポイントは", points)
````
