# 制御構文
- プログラムに電卓以上の高度な機能を持たせるためには、**制御構文**が必要である。  
- **制御構文**とは、プログラムの実行順序を制御するための仕組みである。  
- プログラムの実行は、**順次処理**、**条件分岐（選択）**、**繰り返し（反復・ループ）**の3つの制御構文の組み合わせで表現される。  
- これら3つを組み合わせることで、あらゆる実行順序を記述できる。

----

# 繰り返し（反復、ループ）
- 繰り返し（反復、ループ）には**for文**や **while文**の**繰り返し文** の文法がある。**本稿では、特に重要なfor文を扱います。**

# for文を用いた繰り返し（反復、ループ）
- for文は、繰り返し範囲を指定して処理を行う構文である。範囲を指定するには、**range関数**や**リスト**などのデータの集合（オブジェクト）をin演算子とともに用いる。データの集合のデータを1つずつ変数に代入し、その変数を使って繰り返し処理を行う。**本項ではrange関数のみを扱う。**
  ```python
  for 変数 in オブジェクト:
    実行する処理
  ```

- **注意事項**
- **for文**の処理を記述する際は空白（半角スペース）を入れて字下げしなければならない（これを「インデントを揃える」という）。通常は半角スペース4つ分。半角スペースを毎回4回入力するのは手間なので、Tabキーを押せば自動で調整される。
- **リスト**については次回学習予定のため、

## `range関数`の使い方
- `0`から`n-1`までの`n個`の整数を順番に並べたデータ集合を生成
- 例：range(5) → {0, 1, 2, 3, 4, 5}
- **for文**では0,1,2,3,4を順番に変数に代入して処理を繰り返す(つまり、5回処理を繰り返し)
  ```python
  range(n)
  ```
----
- 開始番号から開始して、 `終了番号-1`までの整数を並べたデータ集合を生成
- 例：range(2, 6) → {2, 3, 4, 5}
  ```python
  range(開始番号, 終了番号)
  ```
----
- 開始番号から開始して、 `終了番号-1`まで、ステップ数の間隔で整数を並べたデータ集合を生成
- 例：range(1, 10, 2) → {1, 3, 5, 7, 9}
  ```python
  range(開始番号, 終了番号, ステップ数)
  ```

# for文を使った演算

## 整数の合計値を計算

**1から100までの`整数`を合計するプログラム**

````python
total = 0                 # 合計を保持する変数で、初期値を0としておく

for i in range(1, 101):  # 終了番号は1加える
    total = total + i    # totalにiを加算

print("合計は", total)    # 合計出力
````

**1から100までの`偶数`を合計するプログラム**

````python
total = 0                   # 合計を保持する変数で、初期値を0としておく

for i in range(2, 101, 2): # 開始番号は2, 終了番号は1加える
    total = total + i      # totalにiを加算

print("合計は ", total)     # 合計出力
````

# for文とif文の組み合わせ

**1から1000までの整数の中で、7の倍数以外の合計を計算するプログラム**

````python
total = 0                      # 合計を保存する変数で、初期値を0としておく

for i in range(1, 1001):      # 終了番号は1加える
    amari = i%7               # あまり計算

    if amari != 0:            # 条件
        total = total + i     # totalにiを加算

print("合計は", total)        # 合計出力
````

**8月に31日間アルバイトをしたとする。1日目から15日目までは日給は3000円、16日目から31日目までは3500円であった場合、8月の合計収入を計算してみよう。for文でカレンダー上の1日から31日まで繰り返すとする。**

````python
total = 0                      # 合計を保存する変数で、初期値を0としておく

for i in range(1, 32):         # カレンダーの1日から31日まで繰り返し(終了番号に1加える)
    if i <= 15:
        total = total + 3000   # 3000を加算
    else:
        total = total + 3500   # 3500を加算

print("合計: ", total, "円")   # 合計出力
````

# 多重for文を使った演算

**掛け算の九九のプログラム（2重for文）**
- 外側のiループと内側のjループを作る
  - i = 1ととき、j = 1, 2, 3, 4, 5, 6, 7, 8, 9と変化
  - i = 2ととき、j = 1, 2, 3, 4, 5, 6, 7, 8, 9と変化
  - ・・・・
  - i = 9ととき、j = 1, 2, 3, 4, 5, 6, 7, 8, 9と変化

````python
for i in range(1, 10):          # 外側ループ：1から9までの数字をiに順に代入
    for j in range(1, 10):      # 内側ループ：1から9までの数字をjに順に代入
        kake = i * j           # iとjの掛け算を計算
        print(kake, end=" ")   # 掛け算の結果を表示し、スペースで区切る（改行しない）
    print()                    # 内側ループ終了後に改行し、次のiの行に移る
````
