# 数値や文字列の画面表示

- 文字列や数値などを画面に出力する関数<br>

# 変数と型
## 変数とは
- 変数は値が出し入れする箱のようなもので、**1つ**の値だけを入れることができる
- 変数に出し入れする値とは**数値**や**文字列**など
- 変数に値を入れることを**代入**という（特に、初めて値を入れる場合は**初期化**という）
- 変数の値は**何度でも入れ替える**ことができる（ただし、古い値は消える）

## 変数の初期化（変数に初めて値を入れる）
- ファイルを閉じなければ、変数の値は以降のセルにも保持される

````python
num1  = 123             # 整数
num2  = 123.456         # 小数
text  = "Hello World!"  # 文字列
````

**プログラムの式は、「左辺 ← 右辺」の操作を意味する。算数や数学の等式とは意味合いが異なることに注意。**

## 変数の画面表示
- 変数にどのような値が代入されているかは、`print関数`で表示できる

**各変数の値を表示**

````python
print(num1)
print(num2)
print(text)
````

**print関数ではコンマで区切って変数を並べることもできる**<br>
表示はスペース区切り

````python
print(num1, num2, text)
````

## 変数の値を変更

````python
num1  = 987            # 整数
num2  = 567.23         # 小数
text  = "See you."     # 文字列

print("num1:", num1)
print("num2:", num2)
print("text:", text)
````

# 算術演算子
プログラム上で演算を行うためには、算術演算子を用いる

| 演算子 | 意味       | 例         | 説明                         |
|--------|------------|------------|------------------------------|
| `+`    | 加算       | `3 + 2`    | 左右の値を足す               |
| `-`    | 減算       | `5 - 1`    | 左から右の値を引く           |
| `*`    | 乗算       | `4 * 3`    | 左右の値を掛ける             |
| `/`    | 除算       | `10 / 2`   | 左を右で割る（結果は浮動小数点数） |
| `//`   | 整数除算   | `10 // 3`  | 商の整数部分（切り捨て）     |
| `%`    | 剰余（余り）| `10 % 3`   | 割り算の余り                 |
| `**`   | 累乗       | `2 ** 3`   | 左の数の右の数乗             |

## 算術演算

````python
# 足し算
tasu = 10 + 3
print("足し算", tasu)

# 引き算
hiku = 10 - 3
print("引き算", hiku)

# 掛け算
kake = 10 * 3
print("掛け算", kake)

# 割り算
waru = 10 / 3
print("割り算", waru)

# 割り算の商
shou = 10 // 3
print("割り算の余り", shou)

# 割り算の余り
amari = 10 % 3
print("割り算の余り", amari)

# べき乗
beki = 10 ** 3
print("3乗", beki)
````

## 変数を使った算術演算
- 同じ操作を繰り返す手間を省くことができ、プログラムの効率や読みやすさが向上する

````python
num1 = 10
num2 = 3

# 足し算
tasu = num1 + num2
print("足し算", tasu)

# 引き算
hiku = num1 - num2
print("引き算", hiku)

# 掛け算
kake = num1 * num2
print("掛け算", kake)

# 割り算
waru = num1 / num2
print("割り算", waru)

# 割り算の商
shou = num1 // num2
print("割り算の余り", shou)

# 割り算の余り
amari = num1 % num2
print("割り算の余り", amari)

# べき乗
beki = num1 ** num2
print("3乗", beki)
````

## 複数の変数の演算
- 直方体の体積を計算するプログラム

````python
edge1 = 10   # 辺1(cm)
edge2 = 20   # 辺2(cm)
edge3 = 30   # 高さ(cm)

volume = edge1 * edge2 * edge3 # 体積の計算

print("体積：", volume, "cm3")
````

## 変数の値をx増やす

- プログラムには以下のような特有の記述方法がある。  
  ```python
  n = n + 1
  ```
- これは、変数 n に現在の値に1を足した結果を再び代入する、という意味である。
- プログラムの式は`「左辺 ← 右辺」の操作`を表し、`数学の等式とは意味が異なる`ことに注意する。

**パターン1**

````python
num1 = 100   # 初期値

num1 = num1 + 2   # 2を足す
print("A", num1) # 表示

num1 = num1 - 2   # 2を引く
print("B", num1) # 表示

num1 = num1 * 2   # 2を掛ける
print("C", num1) # 表示

num1 = num1 / 2   # 2を割る
print("D", num1) # 表示
````

**パターン2（代入演算子を使う）**

| 演算子   | 意味                         | 例                 | 説明                             |
|----------|------------------------------|--------------------|---------------------------------|
| `=`      | 代入                         | `a = 5`            | 右辺の値を左辺の変数に代入する |
| `+=`     | 加算代入                     | `a += 3`           | `a = a + 3` と同じ               |
| `-=`     | 減算代入                     | `a -= 2`           | `a = a - 2` と同じ               |
| `*=`     | 乗算代入                     | `a *= 4`           | `a = a * 4` と同じ               |
| `/=`     | 除算代入                     | `a /= 5`           | `a = a / 5` と同じ               |
| `//=`    | 整数除算代入                 | `a //= 2`          | `a = a // 2` と同じ（切り捨て） |
| `%=`     | 剰余代入                     | `a %= 3`           | `a = a % 3` と同じ               |
| `**=`    | 累乗代入                     | `a **= 2`          | `a = a ** 2` と同じ               |


````python
num2 = 100   # 初期値

num2 += 2   # num2 = num2 + 2と同じ意味
print("A", num2) # 表示

num2 -= 2   # num2 = num2 - 2と同じ意味
print("B",num2) # 表示

num2 *= 2    # num2 = num2 * 2と同じ意味
print("C",num2) # 表示

num2 /= 2   # num2 = num2 / 2と同じ意味
print("D",num2) # 表示

#num1 += 3  # num1 = num1 + 3と同じ意味
````
